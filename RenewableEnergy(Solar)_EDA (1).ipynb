```python
# Exclude the customer with the large 81-day gap
solar_complete = solar_df[solar_df["Customer"] != 2].copy()
# Aggregate across customers for each date and half-hour interval
aggregate_30min = solar_complete[time_cols].sum(axis=0)
print(aggregate_30min)
```

```python
july = solar_complete[
    solar_complete["date"].dt.month == 7
].copy()

july_long = july.melt(
    id_vars=["Customer", "date"],
    value_vars=time_cols,
    var_name="Time",
    value_name="Solar_Generation_kWh"
)

july_long["Time"] = pd.to_datetime(
    july_long["Time"],
    format="%H:%M"
).dt.time

july_long["Timestamp"] = pd.to_datetime(
    july_long["date"].astype(str) + " " +
    july_long["Time"].astype(str)
)

july_aggregate = (
    july_long
    .groupby("Timestamp")["Solar_Generation_kWh"]
    .sum()
    .sort_index()
)

print(july_aggregate.head(10))
print("\nShape:", july_aggregate.shape)
print("\nFrequency check:")
print(july_aggregate.index.to_series().diff().value_counts().head())
```

```python
solar_complete = solar_df[solar_df["Customer"] != 2].copy()

solar_long = solar_complete.melt(
    id_vars=["Customer", "date"],
    value_vars=time_cols,
    var_name="Time",
    value_name="Solar_Generation_kWh"
)

solar_long["Timestamp"] = pd.to_datetime(
    solar_long["date"].astype(str) + " " +
    solar_long["Time"].astype(str)
)

solar_long = solar_long.sort_values("Timestamp")

aggregate_solar = (
    solar_long
    .groupby("Timestamp")["Solar_Generation_kWh"]
    .sum()
    .reset_index()
)

print(aggregate_solar.head())
print("\nShape:", aggregate_solar.shape)
print("\nDate range:")
print(aggregate_solar["Timestamp"].min())
print(aggregate_solar["Timestamp"].max())
```

```python
# Make a copy for feature engineering
solar_ts = aggregate_solar.copy()

# Time-based features
solar_ts["hour"] = solar_ts["Timestamp"].dt.hour
solar_ts["minute"] = solar_ts["Timestamp"].dt.minute
solar_ts["day_of_week"] = solar_ts["Timestamp"].dt.dayofweek
solar_ts["day_of_month"] = solar_ts["Timestamp"].dt.day
solar_ts["month"] = solar_ts["Timestamp"].dt.month
solar_ts["day_of_year"] = solar_ts["Timestamp"].dt.dayofyear
solar_ts["is_weekend"] = (
    solar_ts["day_of_week"] >= 5
).astype(int)

print(solar_ts.head(10))
```

```python
print(solar_ts.dtypes)
```

```python
print("\nUnique values:")
print("Hour:", sorted(solar_ts["hour"].unique()))
print("Minute:", sorted(solar_ts["minute"].unique()))
print("Day of week:", sorted(solar_ts["day_of_week"].unique()))
print("Month:", sorted(solar_ts["month"].unique()))
print("Weekend:", sorted(solar_ts["is_weekend"].unique()))
```

```python
# Lag features
solar_ts["lag_1"] = solar_ts["Solar_Generation_kWh"].shift(1)
solar_ts["lag_2"] = solar_ts["Solar_Generation_kWh"].shift(2)
solar_ts["lag_4"] = solar_ts["Solar_Generation_kWh"].shift(4)
solar_ts["lag_48"] = solar_ts["Solar_Generation_kWh"].shift(48)
solar_ts["lag_96"] = solar_ts["Solar_Generation_kWh"].shift(96)
solar_ts["lag_336"] = solar_ts["Solar_Generation_kWh"].shift(336)

print(solar_ts.head(50))
```

```python
# Past-only rolling features
solar_ts["rolling_mean_2"] = (
    solar_ts["Solar_Generation_kWh"]
    .shift(1)
    .rolling(window=2)
    .mean()
)

solar_ts["rolling_mean_4"] = (
    solar_ts["Solar_Generation_kWh"]
    .shift(1)
    .rolling(window=4)
    .mean()
)

solar_ts["rolling_mean_48"] = (
    solar_ts["Solar_Generation_kWh"]
    .shift(1)
    .rolling(window=48)
    .mean()
)

print(
    solar_ts[
        [
            "Timestamp",
            "Solar_Generation_kWh",
            "rolling_mean_2",
            "rolling_mean_4",
            "rolling_mean_48"
        ]
    ].head(55)
)
```

```python
import numpy as np

solar_ts["hour_sin"] = np.sin(
    2 * np.pi * (solar_ts["hour"] + solar_ts["minute"] / 60) / 24
)
solar_ts["hour_cos"] = np.cos(
    2 * np.pi * (solar_ts["hour"] + solar_ts["minute"] / 60) / 24
)
```

```python
solar_ts["day_of_year_sin"] = np.sin(
    2 * np.pi * solar_ts["day_of_year"] / 365
)

solar_ts["day_of_year_cos"] = np.cos(
    2 * np.pi * solar_ts["day_of_year"] / 365
)
```

```python
solar_ts["is_daylight"] = (
    (solar_ts["hour"] >= 6) &
    (solar_ts["hour"] < 20)
).astype(int)
```

```python
feature_cols = [
    "hour",
    "minute",
    "day_of_week",
    "day_of_month",
    "month",
    "day_of_year",
    "is_weekend",
    "hour_sin",
    "hour_cos",
    "day_of_year_sin",
    "day_of_year_cos",
    "is_daylight",
    "lag_1",
    "lag_2",
    "lag_4",
    "lag_48",
    "lag_96",
    "lag_336",
    "rolling_mean_2",
    "rolling_mean_4",
    "rolling_mean_48"
]

print("Number of features:", len(feature_cols))
print("\nFeature columns:")
print(feature_cols)
```

```python
# Check that every lag/rolling feature uses only past information
print("\nMissing values in engineered features:")
print(solar_ts[feature_cols].isna().sum())
```

```python
print("\nCurrent dataset shape:", solar_ts.shape)
print("\nTarget:")
print("Solar_Generation_kWh")
```

```python
solar_processed = solar_ts.iloc[336:].copy()
print("Shape:", solar_processed.shape)
print("\nRemaining missing values:")
print(solar_processed[feature_cols].isna().sum().sum())
print("\nDate range:")
print(solar_processed["Timestamp"].min())
print(solar_processed["Timestamp"].max())
```

```python
solar_processed = solar_processed.rename(
    columns={
        "Solar_Generation_kWh": "Total_Solar_kWh"
    }
)

final_columns = [
    "Timestamp",
    "Total_Solar_kWh",
    "hour",
    "minute",
    "day_of_week",
    "day_of_month",
    "month",
    "day_of_year",
    "is_weekend",
    "hour_sin",
    "hour_cos",
    "day_of_year_sin",
    "day_of_year_cos",
    "is_daylight",
    "lag_1",
    "lag_2",
    "lag_4",
    "lag_48",
    "lag_96",
    "lag_336",
    "rolling_mean_2",
    "rolling_mean_4",
    "rolling_mean_48"
]
solar_processed = solar_processed[final_columns]
print("Final shape:", solar_processed.shape)
print("\nColumns:")
print(solar_processed.columns.tolist())
print("\nMissing values:")
print(solar_processed.isna().sum().sum())

print("\nFirst 5 rows:")
print(solar_processed.head())
```
